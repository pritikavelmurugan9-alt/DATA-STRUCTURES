```python
class Node:
    def __init__(self,x):
        self.data=x
        self.next=None

class myStack:
    def __init__(self):
        self.top=None
        self.count=0
    def push(self,x):
        temp=Node(x)
        temp.next=self.top
        self.top=temp
        self.count+=1
    def pop(self):
        if self.top is None:
            print("Stack Underflow")
            return -1
        temp=self.top
        self.top=self.top.next
        self.count-=1
        return temp.data

    def peek(self):
        if self.top is None:
            print("Stack is Empty")
            return-1
        return self.top.data

    def isempty(self):
        return self.top is None

    def size(self):
        return self.count

st=myStack()
n=int(input("enter the num of element to push:"))
for i in range(n):
    x=int(input("enter element{i+1}:"))
    st.push(x)

print("\nPopped Element:",st.pop())
print("Top Element:",st.peek())
if st.isempty():
    print("Stack isEmpty")
else:
    print("Stack is not empty")
print("current size:",st.size())
                
                
    
        
```
