```python
stack = []

while True:
    print("\nCurrent Stack:", stack)
    print("1.Push | 2.Pop | 3.Peek | 4.Exit")

    choice = int(input("Enter choice (1-4): "))

    if choice == 1:
        item = input("Enter value to push: ")
        stack.append(item)

    elif choice == 2:
        if stack:
            print("Popped:", stack.pop())
        else:
            print("Stack Underflow! Empty Stack.")

    elif choice == 3:
        if stack:
            print("Top Element:", stack[-1])
        else:
            print("Stack is Empty.")

    elif choice == 4:
        print("Bye Bye!")
        break

    else:
        print("Invalid Choice")
```
